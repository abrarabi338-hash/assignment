# part B
   

```python

Total_Sales = df["Sales"].sum()
print("Total Sales:", Total_Sales)
```

```python
average_sales = df["Sales"].mean()
print("Average Sales/Revenue:", average_sales)
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="Sales", data=df)
plt.title("Sales ")
plt.show()
```

```python
print("Highest category:", df.groupby("Category")["Sales"].sum().idxmax())
```

```python
sns.histplot(df["Category"], bins=20, kde=True)
plt.show()
```

```python
df.hist(figsize=(12,10))
plt.tight_layout()
plt.show()
```

```python
for col in df.columns[:-1]:
    plt.figure(figsize=(5,3))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

```python
Category = df["Category"].value_counts()

plt.pie(
    Category.values,
    labels=Category.index,
    autopct="%1.1f%%"
)

plt.title("Category Distribution")
plt.show()
```

```python
sns.kdeplot(df["Discount"], fill=True)
plt.show()
```

```python
sns.scatterplot(x="Unit_Price", y="Discount", hue="Sales", data=df)
plt.show()
```
