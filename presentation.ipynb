# violin plot

to find distribution of numerical values and find denisity of values

<!-- to find how numeric data is spread -->

```python
sns.violinplot(x="class", y="Alcohol", data=df, color="purple")
plt.show()
```

# scatterplot

to find relation b/w two numerical values

```python
sns.scatterplot(x="Alcohol", y="Malic acid", hue="class", data=df)
plt.show()
```

# find area

to show density of numerical values as smooth curve

```python
sns.kdeplot(df["Alcohol"], fill=True)
plt.show()
```

to find in which range values are more

```python
sns.histplot(df["Alcohol"], bins=20, kde=True)
plt.show()
```

  # to find values in columns

```python
print(df["class"].value_counts())
```

```python

plt.figure(figsize=(10,5))
sns.countplot(x="class", data=df)
plt.title("class Distribution")
plt.show()
```

# pie chart

```python
wine_class = df["class"].value_counts()

plt.pie(
    wine_class.values,
    labels=wine_class.index,
    autopct="%1.1f%%"
)

plt.title("Wine Class Distribution")
plt.show()
```

# most relation 

```python
sns.jointplot(x="Alcohol", y="Malic acid", data=df, kind="hex")
plt.show()
```

# Histograms

to check corelation b/w multiple columns

```python
df.hist(figsize=(12,10))
plt.tight_layout()
plt.show()
```

# Boxplots

to check wether there is outlayers

```python
for col in df.columns[:-1]:
    plt.figure(figsize=(5,3))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

# Correlation Heatmap

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

# Pairplot

to get relations of multime columns at same time 

```python
sns.pairplot(df, hue="class")
plt.show()
```

# Relationship with Outcome

```python
features = [
    "class",
    "Alcohol",
    "Malic acid",
    "Ash",
    "Alcalinity of ash",
    "Magnesium",
    "Total phenols",
    "Flavanoids",
    "Nonflavanoid phenols",
    "Proanthocyanins",
    "Color intensity",
    "Hue",
    "OD280/OD315 of diluted wines",
    "Proline",
    
]
for col in features:
    plt.figure(figsize=(5,4))
    sns.boxplot(x="class", y=col, data=df)
    plt.title(f"{col} vs class")
    plt.show()
```
